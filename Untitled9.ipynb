```python
class Heroi:
    def __init__(self, nome: str, idade: int, tipo: str):
        # Variáveis que guardam as propriedades do herói
        self.nome = nome
        self.idade = idade
        self.tipo = tipo.lower()  # Padroniza para letras minúsculas

    # Função (método) que executa a ação de ataque
    def atacar(self):
        # Estrutura de decisão para definir o ataque com base no tipo
        if self.tipo == "mago":
            ataque = "usou magia"
        elif self.tipo == "guerreiro":
            ataque = "usou espada"
        elif self.tipo == "monge":
            ataque = "usou artes marciais"
        elif self.tipo == "ninja":
            ataque = "usou shuriken"
        else:
            ataque = "usou um ataque indefinido"

        # Exibe a mensagem final concatenando o tipo e o ataque
        print(f"O {self.tipo} atacou usando {ataque}")

# Criando os objetos (instâncias da classe)
mago = Heroi("Gandalf", 2019, "mago")
guerreiro = Heroi("Aragorn", 87, "guerreiro")
monge = Heroi("Aang", 112, "monge")
ninja = Heroi("Naruto", 17, "ninja")


mago.atacar()
guerreiro.atacar()
monge.atacar()
ninja.atacar()
```
